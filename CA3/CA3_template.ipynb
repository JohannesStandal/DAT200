### Imports

```python
import sklearn as sk
import seaborn as sns
import pandas as pd
```

### Reading data

```python
df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

df_train.shape
```

### Data exploration and visualisation

```python
df_visual = df_train.drop('Id', axis=1)

sns.pairplot(df_visual,  hue='quality', plot_kws={'s':5})
```

### Data cleaning

```python
df_train.dropna(axis=0)
```

### Data preprocessing and visualisation

```python
x_train = df_train.drop(['quality', 'Id'], axis=1)
y_train = df_train['quality']

X_train, X_test, Y_train, Y_test = sk.model_selection.train_test_split(x_train, y_train, test_size=0.2, random_state=42)

X_train_scaled = sk.preprocessing.StandardScaler().fit_transform(X_train)
X_test_scaled = sk.preprocessing.StandardScaler().fit_transform(X_test)

test = df_test.drop('Id', axis=1)
test_scaled = sk.preprocessing.StandardScaler().fit_transform(test)
id_array = df_test['Id'].to_numpy()

sns.violinplot(X_train_scaled)
```

### Modelling

```python
model = sk.linear_model.LogisticRegression(max_iter=2000, random_state=42)
model.fit(X_train_scaled, Y_train)

y_pred = model.predict(X_test_scaled)

y_probs = model.predict_proba(X_test_scaled)[:, 1] # Sannsynlighet for klasse 1

# 6. Evaluer modellen
print(f"Nøyaktighet (Accuracy): {sk.metrics.accuracy_score(Y_test, y_pred):.2f}")
print("\nKlassifiseringsrapport:\n", sk.metrics.classification_report(Y_test, y_pred))


```
