```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# Loading the data

df = pd.read_csv('starbucks.csv', sep=',', header=0, encoding='latin1')

# Visualising the data using pairwise scatter-plots

sns.pairplot(df)
plt.tight_layout()
plt.show()
```

```python
# Defining the predictor matrix and the response variable 

X=df[['fat', 'carb','fiber','protein']]

y=df['calories']

# Add feature 1 to the data
X = np.c_[np.ones(X.shape[0]), X]

# Least squares solution
w = np.dot(np.linalg.inv(np.dot(X.T, X)), np.dot(X.T, y))

print(w)

# Predicted values
y_pred = np.dot(X, w)

# Plot regression solution and data
plt.title("Least squares fit: y = {:.3f} + {:.3f}fat + {:.3f}carb + {:.3f}fiber + {:.3f}protein".format(w[0], w[1], w[2], w[3], w[4]))
plt.scatter(y, y_pred)
plt.xlabel("Calories")
plt.ylabel("Predicted calories")
plt.show()
```
